```python
DECISION_PROMPT = """
You are the decision engine for ACE Helper.

Analyze the customer's message and choose exactly ONE action.

ANSWER:
Use when ACE can answer using general customer-support knowledge.

ASK_INFORMATION:
Use when the customer needs to provide missing information before ACE can proceed.

USE_TOOL:
Use when the question requires checking real customer-specific data from a database, API, order system, payment system, account system, delivery system, or another external system.

ESCALATE:
Use when the customer explicitly asks for a human agent or the issue requires human intervention.

IMPORTANT:
- Questions about a specific order, order status, delivery status, payment, refund, transaction, account details, or personal customer data require USE_TOOL unless the required information is already available from a tool.
- "Where is my order?" means USE_TOOL.
- "What is my order number?" can be ANSWER if the order number was already provided by the customer in the conversation.
- "What are your support hours?" means ANSWER.
- "How do I return a product?" means ANSWER if the return policy is available in ACE's knowledge.
- "I want to speak to a human agent." means ESCALATE.

Return ONLY one of:
ANSWER
ASK_INFORMATION
USE_TOOL
ESCALATE
"""
```

```python
from langchain_ollama import ChatOllama

llm_cpu = ChatOllama(
    model="llama3.2",
    temperature=0.2,
    num_gpu=0
)
```

```python
decision_prompt = ChatPromptTemplate.from_messages([
    ("system", DECISION_PROMPT),
    ("human", "{customer_message}")
])

decision_chain = decision_prompt | llm_cpu
```

```python
def clean_decision(result):
    decision = result.content.strip().upper()

    for value in ["ANSWER", "ASK_INFORMATION", "USE_TOOL", "ESCALATE"]:
        if value in decision:
            return value

    return "ANSWER"
    
```

```python
orders = {
    "12345": {
        "customer_name": "Nikhil",
        "status": "Shipped",
        "estimated_delivery": "28 September 2026",
        "product": "Laptop"
    },
    "12346": {
        "customer_name": "Rahul",
        "status": "Processing",
        "estimated_delivery": "30 September 2026",
        "product": "Headphones"
    }
}
```

```python
def get_order_status(order_number):
    order = orders.get(str(order_number))

    if not order:
        return {
            "found": False,
            "message": f"No order found for order number {order_number}."
        }

    return {
        "found": True,
        "order_number": str(order_number),
        **order
    }
```

```python
from langchain_core.tools import tool

@tool
def order_lookup(order_number: str) -> dict:
    """Look up customer order information using the order number."""
    return get_order_status(order_number)
```

```python
def ace_order_response(customer_message):
    # 1. Ask decision engine what to do
    decision_response = decision_chain.invoke({
        "customer_message": customer_message
    })

    decision = clean_decision(decision_response)

    print("Decision:", decision)

    # 2. If order lookup is required
    if decision == "USE_TOOL":

        # Simple order-number extraction
        import re

        match = re.search(r"\b\d{5,}\b", customer_message)

        if not match:
            return "Please provide your order number so I can check your order status."

        order_number = match.group()

        # 3. Call the LangChain tool
        order_result = order_lookup.invoke({
            "order_number": order_number
        })

        # 4. Give tool result to LLM
        response = llm_cpu.invoke(
            f"""
You are ACE Helper.

Customer message:
{customer_message}

Verified order information from the company system:
{order_result}

Answer the customer naturally and professionally.

Important:
- Use only the verified order information above.
- Do not invent anything.
- Do not mention internal tools or decision engines.
"""
        )

        return response.content

    return "This request does not require an order lookup."
```

```python
import re

def get_decision(message):
    text = message.lower()

    # Human escalation
    if user_requests_human(message):
        return "ESCALATE"

    # Customer-specific order request
    order_words = [
        "order",
        "delivery",
        "delivered",
        "shipment",
        "shipped",
        "arrive"
    ]

    if any(word in text for word in order_words):
        # Order number present
        if re.search(r"\b\d{5,}\b", text):
            return "USE_TOOL"

        # Order mentioned but number missing
        return "ASK_INFORMATION"

    # General question
    return "ANSWER"
```

```python
def handle_order_request(message):
    import re

    match = re.search(r"\b\d{5,}\b", message)

    if not match:
        return "Please provide your order number so I can check it."

    order_number = match.group()

    result = order_lookup.invoke({
        "order_number": order_number
    })

    return result
```

```python
company_knowledge = [
    {
        "title": "Return Policy",
        "content": "Customers can request a return within 7 days of delivery. The product must be unused and in its original packaging."
    },
    {
        "title": "Refund Policy",
        "content": "Approved refunds are processed within 5-7 business days after the returned product is received and inspected."
    },
    {
        "title": "Shipping Policy",
        "content": "Standard delivery usually takes 3-5 business days. Delivery times may vary depending on the customer's location."
    },
    {
        "title": "Cancellation Policy",
        "content": "Orders can be cancelled before they are shipped. Once an order has been shipped, cancellation may not be possible."
    },
    {
        "title": "Support Hours",
        "content": "Customer support is available Monday to Saturday from 9:00 AM to 6:00 PM."
    }
]

print(f"Knowledge base created: {len(company_knowledge)} documents")
```

```python
def search_knowledge(query):
    query_words = set(query.lower().split())

    results = []

    for doc in company_knowledge:
        content = (doc["title"] + " " + doc["content"]).lower()
        score = sum(word in content for word in query_words)

        if score > 0:
            results.append((score, doc))

    results.sort(key=lambda x: x[0], reverse=True)

    return [doc for score, doc in results[:2]]
```

```python
results = search_knowledge("What is your return policy?")

for doc in results:
    print(doc["title"])
    print(doc["content"])
    print()
```

```python
def generate_knowledge_response(customer_message):
    results = search_knowledge(customer_message)

    if not results:
        return "I couldn't find this information in our company knowledge base."

    context = "\n\n".join(
        f"{doc['title']}:\n{doc['content']}"
        for doc in results
    )

    response = llm_cpu.invoke(f"""
You are ACE Helper, a customer support assistant.

Customer question:
{customer_message}

Company knowledge:
{context}

Answer the customer using ONLY the company knowledge provided above.

Rules:
- Do not invent policies or information.
- If the answer is not available, say that you don't have that information.
- Keep the answer clear and concise.
""")

    return response.content
```

```python
def get_order_from_history():
    for message in reversed(chat_history):
        match = re.search(r"\b\d{5,}\b", message.content)

        if match:
            return match.group()

    return None
```

```python
import re
import uuid
from datetime import datetime
import ipywidgets as widgets
from IPython.display import display, clear_output


class ACEApp:

    def __init__(self, llm, knowledge_base, order_data):
        self.llm = llm
        self.knowledge_base = knowledge_base
        self.orders = order_data

        self.sessions = {}
        self.tickets = {}

        self.build_ui()


                                                              
                        
                                                              

    def get_session(self, session_id):

        if session_id not in self.sessions:
            self.sessions[session_id] = {
                "messages": [],
                "active_ticket": None
            }

        return self.sessions[session_id]


    def save_session_message(self, session_id, sender, message):

        session = self.get_session(session_id)

        session["messages"].append({
            "sender": sender,
            "message": message,
            "time": datetime.now().strftime("%H:%M:%S")
        })


                                                              
                      
                                                              

    def search_knowledge(self, query):

        text = query.lower()

        topic_words = {
            "return": [
                "return",
                "returning",
                "exchange"
            ],
            "refund": [
                "refund",
                "money back"
            ],
            "shipping": [
                "shipping",
                "delivery",
                "deliver",
                "shipping time"
            ],
            "cancellation": [
                "cancel",
                "cancellation"
            ],
            "support": [
                "support hours",
                "support timing",
                "working hours",
                "customer support"
            ]
        }

        matched_topics = []

        for topic, words in topic_words.items():

            if any(word in text for word in words):
                matched_topics.append(topic)

        results = []

        for doc in self.knowledge_base:

            doc_text = (
                doc["title"] + " " + doc["content"]
            ).lower()

            for topic in matched_topics:

                if topic in doc_text:
                    results.append(doc)
                    break

        return results[:2]


                                                              
                  
                                                              

    def find_order_number(self, text):

        match = re.search(
            r"\b\d{5,}\b",
            text
        )

        if match:
            return match.group()

        return None


    def get_previous_order_number(self, session_id):

        session = self.get_session(session_id)

        for message in reversed(session["messages"]):

            number = self.find_order_number(
                message["message"]
            )

            if number:
                return number

        return None


    def lookup_order(self, order_number):

        order = self.orders.get(
            str(order_number)
        )

        if not order:

            return {
                "found": False,
                "order_number": str(order_number)
            }

        return {
            "found": True,
            "order_number": str(order_number),
            **order
        }


                                                              
                   
                                                              

    def wants_human(self, message):

        phrases = [
            "human agent",
            "speak to a human",
            "talk to a human",
            "live agent",
            "real person",
            "human support",
            "customer support agent",
            "representative"
        ]

        text = message.lower()

        return any(
            phrase in text
            for phrase in phrases
        )


                                                              
                   
                                                              

    def create_ticket(
        self,
        session_id,
        initial_message
    ):

        ticket_id = (
            "ACE-" +
            uuid.uuid4().hex[:6].upper()
        )

        session = self.get_session(session_id)

        ticket = {
            "ticket_id": ticket_id,
            "session_id": session_id,
            "status": "OPEN",
            "assigned_to": "Nikhil",
            "created_at": datetime.now().strftime(
                "%Y-%m-%d %H:%M:%S"
            ),
            "messages": []
        }

                                   
        for msg in session["messages"]:

            ticket["messages"].append({
                "sender": msg["sender"],
                "message": msg["message"],
                "time": msg["time"]
            })

                                                  
        if not ticket["messages"] or ticket["messages"][-1]["message"] != initial_message:

            ticket["messages"].append({
                "sender": "customer",
                "message": initial_message,
                "time": datetime.now().strftime("%H:%M:%S")
            })

        self.tickets[ticket_id] = ticket

        session["active_ticket"] = ticket_id

        return ticket


                                                              
                        
                                                              

    def add_ticket_message(
        self,
        ticket_id,
        sender,
        message
    ):

        ticket = self.tickets.get(ticket_id)

        if not ticket:
            return False

        ticket["messages"].append({
            "sender": sender,
            "message": message,
            "time": datetime.now().strftime("%H:%M:%S")
        })

        return True


                                                              
                       
                                                              

    def order_response(
        self,
        customer_message,
        order_number
    ):

        result = self.lookup_order(
            order_number
        )

        prompt = f"""
You are ACE Helper.

Customer:
{customer_message}

Verified order information:
{result}

Reply to the customer naturally.

Rules:
- Use only the verified information.
- Do not invent information.
- Do not mention internal tools.
- Keep the response concise.
"""

        return self.llm.invoke(
            prompt
        ).content


                                                              
                           
                                                              

    def knowledge_response(
        self,
        customer_message,
        documents
    ):

        context = "\n\n".join(
            f"{doc['title']}:\n{doc['content']}"
            for doc in documents
        )

        prompt = f"""
You are ACE Helper.

Customer question:
{customer_message}

Verified company information:
{context}

Answer the customer using ONLY the information above.

Rules:
- Do not invent policies.
- Do not add unsupported information.
- Keep the response concise and professional.
"""

        return self.llm.invoke(
            prompt
        ).content


                                                              
                         
                                                              

    def general_ai_response(
        self,
        session_id,
        customer_message
    ):

        session = self.get_session(
            session_id
        )

        history = session["messages"][-10:]

        conversation = "\n".join(
            f"{m['sender'].upper()}: {m['message']}"
            for m in history
        )

        prompt = f"""
You are ACE Helper, an AI customer support assistant.

Conversation:
{conversation}

Latest customer message:
{customer_message}

Respond professionally and concisely.

Rules:
- Never invent company policies.
- Never invent order information.
- If you do not know something, say so.
- The customer can request a human agent if needed.
"""

        return self.llm.invoke(
            prompt
        ).content


                                                              
                    
                                                              

    def process_customer_message(
        self,
        session_id,
        message
    ):

        message = message.strip()

        if not message:
            return

        session = self.get_session(
            session_id
        )

                                              
                              
                                              

        active_ticket_id = session["active_ticket"]

        if active_ticket_id:

            ticket = self.tickets.get(
                active_ticket_id
            )

            if ticket and ticket["status"] != "RESOLVED":

                self.save_session_message(
                    session_id,
                    "customer",
                    message
                )

                self.add_ticket_message(
                    active_ticket_id,
                    "customer",
                    message
                )

                ticket["status"] = "IN_PROGRESS"

                self.render_customer_chat()
                self.refresh_ticket_list()

                return


                                              
                               
                                              

        self.save_session_message(
            session_id,
            "customer",
            message
        )


                                              
                          
                                              

        if self.wants_human(message):

            ticket = self.create_ticket(
                session_id,
                message
            )

            response = (
                "I've escalated your request to a "
                f"human support agent.\n\n"
                f"Your ticket ID is {ticket['ticket_id']}."
            )

            self.save_session_message(
                session_id,
                "ai",
                response
            )

            self.add_ticket_message(
                ticket["ticket_id"],
                "ai",
                response
            )

            self.render_customer_chat()
            self.refresh_ticket_list()

            self.ticket_dropdown.value = (
                ticket["ticket_id"]
            )

            return


                                              
                       
                                              

        order_words = [
            "order",
            "delivery",
            "delivered",
            "shipment",
            "shipped",
            "arrive"
        ]

        if any(
            word in message.lower()
            for word in order_words
        ):

            order_number = (
                self.find_order_number(message)
            )

            if not order_number:

                order_number = (
                    self.get_previous_order_number(
                        session_id
                    )
                )

            if not order_number:

                response = (
                    "Sure. Please provide your "
                    "order number so I can check it."
                )

            else:

                response = self.order_response(
                    message,
                    order_number
                )

            self.save_session_message(
                session_id,
                "ai",
                response
            )

            self.render_customer_chat()

            return


                                              
                           
                                              

        documents = self.search_knowledge(
            message
        )

        if documents:

            response = self.knowledge_response(
                message,
                documents
            )

            self.save_session_message(
                session_id,
                "ai",
                response
            )

            self.render_customer_chat()

            return


                                              
                    
                                              

        response = self.general_ai_response(
            session_id,
            message
        )

        self.save_session_message(
            session_id,
            "ai",
            response
        )

        self.render_customer_chat()


                                                              
                 
                                                              

    def customer_send_handler(self, button):

        message = (
            self.customer_input.value.strip()
        )

        if not message:
            return

        self.customer_input.value = ""

        self.process_customer_message(
            "customer_001",
            message
        )


    def new_customer_chat(self, button):

        self.sessions["customer_001"] = {
            "messages": [],
            "active_ticket": None
        }

        self.render_customer_chat()


    def render_customer_chat(self):

        session = self.get_session(
            "customer_001"
        )

        with self.customer_output:

            clear_output(wait=True)

            print("🤖 ACE HELPER")
            print("=" * 70)

            if not session["messages"]:

                print(
                    "Welcome to ACE Helper. "
                    "How can I help you today?"
                )

                return

            for msg in session["messages"]:

                if msg["sender"] == "customer":
                    label = "YOU"

                elif msg["sender"] == "ai":
                    label = "ACE HELPER"

                else:
                    label = "HUMAN AGENT"

                print(
                    f"\n{label}:"
                )

                print(
                    msg["message"]
                )

                print("-" * 70)


                                                              
                     
                                                              

    def refresh_ticket_list(self, button=None):

        options = []

        for ticket_id, ticket in self.tickets.items():

            if ticket["status"] != "RESOLVED":

                label = (
                    f"{ticket_id} | "
                    f"{ticket['status']}"
                )

                options.append(
                    (label, ticket_id)
                )

        self.ticket_dropdown.options = options

        if options:

            if self.ticket_dropdown.value not in [
                value for label, value in options
            ]:

                self.ticket_dropdown.value = (
                    options[0][1]
                )

            self.show_selected_ticket(
                self.ticket_dropdown.value
            )

        else:

            with self.agent_conversation:

                clear_output()

                print(
                    "No open tickets."
                )


    def show_selected_ticket(
        self,
        ticket_id
    ):

        if not ticket_id:
            return

        ticket = self.tickets.get(
            ticket_id
        )

        if not ticket:
            return

        with self.agent_conversation:

            clear_output(wait=True)

            print(
                f"Ticket: {ticket_id}"
            )

            print(
                f"Status: {ticket['status']}"
            )

            print(
                f"Assigned: {ticket['assigned_to']}"
            )

            print("=" * 70)

            for msg in ticket["messages"]:

                print(
                    f"\n{msg['sender'].upper()}:"
                )

                print(
                    msg["message"]
                )

                print("-" * 70)


                                                              
                     
                                                              

    def agent_assist_handler(self, button):

        ticket_id = (
            self.ticket_dropdown.value
        )

        if not ticket_id:
            return

        ticket = self.tickets.get(
            ticket_id
        )

        if not ticket:
            return

        conversation = "\n".join(
            f"{m['sender'].upper()}: {m['message']}"
            for m in ticket["messages"]
        )

        knowledge = "\n".join(
            f"{d['title']}: {d['content']}"
            for d in self.knowledge_base
        )

        prompt = f"""
You are ACE Helper AI Agent Assist.

Human agent is handling this ticket.

Conversation:
{conversation}

Company knowledge:
{knowledge}

Provide:

1. Issue summary
2. Relevant company information
3. Suggested response

Rules:
- Do not invent information.
- Do not promise compensation.
- Human agent has final authority.
- Keep it concise.
"""

        with self.ai_assist_output:

            clear_output()

            print(
                "Generating AI suggestion..."
            )

        response = self.llm.invoke(
            prompt
        ).content

        with self.ai_assist_output:

            clear_output()

            print(" AI AGENT ASSIST")
            print("=" * 70)
            print(response)


                                                              
                
                                                              

    def agent_send_handler(self, button):

        ticket_id = (
            self.ticket_dropdown.value
        )

        message = (
            self.agent_input.value.strip()
        )

        if not ticket_id or not message:
            return

        ticket = self.tickets.get(
            ticket_id
        )

        if not ticket:
            return

        self.add_ticket_message(
            ticket_id,
            "agent",
            message
        )

        ticket["status"] = (
            "WAITING_FOR_CUSTOMER"
        )

        session = self.get_session(
            ticket["session_id"]
        )

        self.save_session_message(
            ticket["session_id"],
            "agent",
            message
        )

        self.agent_input.value = ""

        self.show_selected_ticket(
            ticket_id
        )

        self.render_customer_chat()

        self.refresh_ticket_list()


                                                              
             
                                                              

    def resolve_ticket_handler(self, button):

        ticket_id = (
            self.ticket_dropdown.value
        )

        if not ticket_id:
            return

        ticket = self.tickets.get(
            ticket_id
        )

        if not ticket:
            return

        ticket["status"] = "RESOLVED"

        self.add_ticket_message(
            ticket_id,
            "system",
            "Ticket resolved by human agent."
        )

        session = self.get_session(
            ticket["session_id"]
        )

        session["active_ticket"] = None

        self.refresh_ticket_list()


                                                              
              
                                                              

    def build_ui(self):

                                                    

        self.customer_title = widgets.HTML(
            "<h2>ACE Helper — Customer</h2>"
        )

        self.customer_output = widgets.Output(
            layout=widgets.Layout(
                border="1px solid #ccc",
                padding="12px",
                height="380px",
                overflow="auto"
            )
        )

        self.customer_input = widgets.Textarea(
            placeholder="Type your message...",
            layout=widgets.Layout(
                width="100%",
                height="70px"
            )
        )

        self.customer_send_button = widgets.Button(
            description="Send",
            button_style="primary"
        )

        self.new_chat_button = widgets.Button(
            description="New Chat",
            button_style="warning"
        )


                                                 

        self.agent_title = widgets.HTML(
            "<h2> Human Agent Dashboard</h2>"
        )

        self.refresh_button = widgets.Button(
            description="Refresh Tickets",
            button_style="info"
        )

        self.ticket_dropdown = widgets.Dropdown(
            description="Ticket:",
            options=[],
            layout=widgets.Layout(
                width="100%"
            )
        )

        self.agent_conversation = widgets.Output(
            layout=widgets.Layout(
                border="1px solid #ccc",
                padding="12px",
                height="320px",
                overflow="auto"
            )
        )

        self.ai_assist_button = widgets.Button(
            description="AI Agent Assist",
            button_style="info"
        )

        self.ai_assist_output = widgets.Output(
            layout=widgets.Layout(
                border="1px solid #ccc",
                padding="12px",
                height="200px",
                overflow="auto"
            )
        )

        self.agent_input = widgets.Textarea(
            placeholder="Reply to customer...",
            layout=widgets.Layout(
                width="100%",
                height="70px"
            )
        )

        self.agent_send_button = widgets.Button(
            description="Send to Customer",
            button_style="success"
        )

        self.resolve_button = widgets.Button(
            description="Resolve Ticket",
            button_style="warning"
        )


                                                  

        self.customer_send_button.on_click(
            self.customer_send_handler
        )

        self.new_chat_button.on_click(
            self.new_customer_chat
        )

        self.refresh_button.on_click(
            self.refresh_ticket_list
        )

        self.ticket_dropdown.observe(
            self.ticket_dropdown_changed,
            names="value"
        )

        self.ai_assist_button.on_click(
            self.agent_assist_handler
        )

        self.agent_send_button.on_click(
            self.agent_send_handler
        )

        self.resolve_button.on_click(
            self.resolve_ticket_handler
        )


                                                  

        customer_panel = widgets.VBox([

            self.customer_title,

            self.customer_output,

            self.customer_input,

            widgets.HBox([
                self.customer_send_button,
                self.new_chat_button
            ])
        ])


        agent_panel = widgets.VBox([

            self.agent_title,

            self.refresh_button,

            self.ticket_dropdown,

            self.agent_conversation,

            self.ai_assist_button,

            self.ai_assist_output,

            self.agent_input,

            widgets.HBox([
                self.agent_send_button,
                self.resolve_button
            ])
        ])


        self.ui = widgets.VBox([

            customer_panel,

            widgets.HTML(
                "<hr><h2>Human Agent Section</h2>"
            ),

            agent_panel
        ])


    def ticket_dropdown_changed(self, change):

        if change["name"] == "value":

            self.show_selected_ticket(
                change["new"]
            )


                                                              
               
                                                              

    def start(self):

        display(self.ui)

        self.render_customer_chat()

        self.refresh_ticket_list()

        print(
            "ACE HELPER V1 STARTED"
        )

        print(
            "Customer → AI → RAG/Tool → "
            "Human Agent → Customer"
        )


                                                              
                 
                                                              

ace_app = ACEApp(
    llm=llm_cpu,
    knowledge_base=company_knowledge,
    order_data=orders
)

ace_app.start()
```

```python
import sqlite3
from datetime import datetime

DB_PATH = "ace_helper.db"

def db_execute(query, params=(), fetch=False, many=False):
    conn = sqlite3.connect(DB_PATH)
    cursor = conn.cursor()

    if many:
        cursor.executemany(query, params)
    else:
        cursor.execute(query, params)

    result = cursor.fetchall() if fetch else None
    conn.commit()
    conn.close()
    return result


def save_ticket_to_db(ticket):
    db_execute(
        """
        INSERT OR REPLACE INTO tickets
        (ticket_id, conversation_id, status, assigned_to, created_at)
        VALUES (?, ?, ?, ?, ?)
        """,
        (
            ticket["ticket_id"],
            ticket["session_id"],
            ticket["status"],
            ticket["assigned_to"],
            ticket["created_at"]
        )
    )


def save_ticket_message_to_db(ticket_id, sender, message, created_at=None):
    if created_at is None:
        created_at = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    db_execute(
        """
        INSERT INTO ticket_messages
        (ticket_id, sender, message, created_at)
        VALUES (?, ?, ?, ?)
        """,
        (ticket_id, sender, message, created_at)
    )


def save_customer_to_db(customer_id, name=""):
    db_execute(
        """
        INSERT OR IGNORE INTO customers
        (customer_id, name)
        VALUES (?, ?)
        """,
        (customer_id, name)
    )


def save_conversation_to_db(conversation_id, customer_id):
    db_execute(
        """
        INSERT OR IGNORE INTO conversations
        (conversation_id, customer_id, created_at)
        VALUES (?, ?, ?)
        """,
        (
            conversation_id,
            customer_id,
            datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        )
    )


print("Database functions connected.")
```

```python
import types

def db_create_ticket(self, session_id, initial_message):
    ticket_id = "ACE-" + uuid.uuid4().hex[:6].upper()
    session = self.get_session(session_id)

    created_at = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    ticket = {
        "ticket_id": ticket_id,
        "session_id": session_id,
        "status": "OPEN",
        "assigned_to": "Nikhil",
        "created_at": created_at,
        "messages": []
    }

    for msg in session["messages"]:
        ticket["messages"].append({
            "sender": msg["sender"],
            "message": msg["message"],
            "time": msg["time"]
        })

    if not ticket["messages"] or ticket["messages"][-1]["message"] != initial_message:
        ticket["messages"].append({
            "sender": "customer",
            "message": initial_message,
            "time": datetime.now().strftime("%H:%M:%S")
        })

    self.tickets[ticket_id] = ticket
    session["active_ticket"] = ticket_id

    save_ticket_to_db(ticket)

    for msg in ticket["messages"]:
        save_ticket_message_to_db(
            ticket_id,
            msg["sender"],
            msg["message"],
            created_at
        )

    save_customer_to_db(session_id, "Nikhil")
    save_conversation_to_db(session_id, session_id)

    return ticket


def db_add_ticket_message(self, ticket_id, sender, message):
    ticket = self.tickets.get(ticket_id)

    if not ticket:
        return False

    time_now = datetime.now().strftime("%H:%M:%S")

    ticket["messages"].append({
        "sender": sender,
        "message": message,
        "time": time_now
    })

    save_ticket_message_to_db(
        ticket_id,
        sender,
        message
    )

    save_ticket_to_db(ticket)

    return True


ace_app.create_ticket = types.MethodType(
    db_create_ticket,
    ace_app
)

ace_app.add_ticket_message = types.MethodType(
    db_add_ticket_message,
    ace_app
)

print("Ticket database integration enabled.")
```

```python
def db_process_customer_message(self, session_id, message):
    message = message.strip()

    if not message:
        return

    session = self.get_session(session_id)
    active_ticket_id = session["active_ticket"]

    if active_ticket_id:
        ticket = self.tickets.get(active_ticket_id)

        if ticket and ticket["status"] != "RESOLVED":
            self.save_session_message(
                session_id,
                "customer",
                message
            )

            self.add_ticket_message(
                active_ticket_id,
                "customer",
                message
            )

            self.set_ticket_status(
                active_ticket_id,
                "IN_PROGRESS"
            )

            self.render_customer_chat()
            self.refresh_ticket_list()
            return

    self.save_session_message(
        session_id,
        "customer",
        message
    )

    if self.wants_human(message):
        ticket = self.create_ticket(
            session_id,
            message
        )

        response = (
            "I've escalated your request to a "
            f"human support agent.\n\n"
            f"Your ticket ID is {ticket['ticket_id']}."
        )

        self.save_session_message(
            session_id,
            "ai",
            response
        )

        self.add_ticket_message(
            ticket["ticket_id"],
            "ai",
            response
        )

        self.render_customer_chat()
        self.refresh_ticket_list()

        self.ticket_dropdown.value = ticket["ticket_id"]
        return

    if self.find_order_number(message):
        order_number = self.find_order_number(message)
        result = self.lookup_order(order_number)

        if result:
            response = (
                f"Order {order_number} status: "
                f"{result['status']}. "
                f"Estimated delivery: "
                f"{result['estimated_delivery']}."
            )
        else:
            response = f"I couldn't find order {order_number}."

        self.save_session_message(
            session_id,
            "ai",
            response
        )

        self.render_customer_chat()
        return

    knowledge = self.search_knowledge(message)

    if knowledge:
        context = "\n\n".join(
            item["content"]
            for item in knowledge
        )

        prompt = f"""
You are ACE Helper, a customer support assistant.

Use only the following company information:

{context}

Customer message:
{message}

Answer clearly and concisely.
Do not invent company policies.
"""

        response = self.llm.invoke(prompt).content

    else:
        response = self.llm.invoke(message).content

    self.save_session_message(
        session_id,
        "ai",
        response
    )

    self.render_customer_chat()


ace_app.process_customer_message = types.MethodType(
    db_process_customer_message,
    ace_app
)

print("Customer ticket status persistence enabled.")
```

```python
import requests
import ipywidgets as widgets
from IPython.display import display

ticket_dropdown = widgets.Dropdown(
    description="Ticket:",
    options=[]
)

conversation_box = widgets.Output(
    layout=widgets.Layout(
        border="1px solid #ccc",
        padding="10px",
        height="300px",
        overflow="auto"
    )
)

agent_input = widgets.Textarea(
    placeholder="Reply to customer...",
    layout=widgets.Layout(width="100%", height="70px")
)

send_button = widgets.Button(
    description="Send to Customer",
    button_style="success"
)


def load_tickets():
    response = requests.get(
        "http://127.0.0.1:8000/tickets"
    )

    data = response.json()

    ticket_dropdown.options = [
        (
            f"{t['ticket_id']} — {t['status']}",
            t["ticket_id"]
        )
        for t in data["tickets"]
    ]


def load_conversation(change=None):

    ticket_id = ticket_dropdown.value

    if not ticket_id:
        return

    response = requests.get(
        f"http://127.0.0.1:8000/tickets/{ticket_id}/conversation"
    )

    data = response.json()

    conversation_box.clear_output(wait=True)

    with conversation_box:

        print(f"Ticket: {ticket_id}")
        print("=" * 60)

        for msg in data["messages"]:

            print(
                f"{msg['sender'].upper()}: "
                f"{msg['message']}"
            )

            print(
                f"Time: {msg['created_at']}"
            )

            print("-" * 60)


def send_message(button):

    ticket_id = ticket_dropdown.value
    message = agent_input.value.strip()

    if not ticket_id:
        print("Select a ticket first.")
        return

    if not message:
        print("Enter a message first.")
        return

    response = requests.post(
        f"http://127.0.0.1:8000/tickets/{ticket_id}/reply",
        json={
            "message": message
        }
    )

    print("Status:", response.status_code)

    if response.status_code == 200:

        agent_input.value = ""

        load_conversation()

    else:

        print(response.text)


ticket_dropdown.observe(
    load_conversation,
    names="value"
)

send_button.on_click(send_message)

load_tickets()

display(
    widgets.HTML(
        "<h2>Human Agent Dashboard</h2>"
    ),
    ticket_dropdown,
    conversation_box,
    agent_input,
    send_button
)
```

```python
import requests
import ipywidgets as widgets
from IPython.display import display


ticket_dropdown = widgets.Dropdown(
    description="Ticket:",
    options=[]
)

conversation_box = widgets.Output(
    layout=widgets.Layout(
        border="1px solid #ccc",
        padding="10px",
        height="300px",
        overflow="auto"
    )
)

agent_input = widgets.Textarea(
    placeholder="Reply to customer...",
    layout=widgets.Layout(
        width="100%",
        height="70px"
    )
)

send_button = widgets.Button(
    description="Send to Customer",
    button_style="success"
)

resolve_button = widgets.Button(
    description="Resolve Ticket",
    button_style="danger"
)


def load_tickets():

    response = requests.get(
        "http://127.0.0.1:8000/tickets"
    )

    data = response.json()

    ticket_dropdown.options = [
        (
            f"{t['ticket_id']} — {t['status']}",
            t["ticket_id"]
        )
        for t in data["tickets"]
    ]


def load_conversation(change=None):

    ticket_id = ticket_dropdown.value

    if not ticket_id:
        return

    response = requests.get(
        f"http://127.0.0.1:8000/tickets/{ticket_id}/conversation"
    )

    data = response.json()

    conversation_box.clear_output(wait=True)

    with conversation_box:

        print(f"Ticket: {ticket_id}")
        print("=" * 60)

        for msg in data["messages"]:

            print(
                f"{msg['sender'].upper()}: "
                f"{msg['message']}"
            )

            print(
                f"Time: {msg['created_at']}"
            )

            print("-" * 60)


def send_message(button):

    ticket_id = ticket_dropdown.value
    message = agent_input.value.strip()

    if not ticket_id:
        print("Select a ticket first.")
        return

    if not message:
        print("Enter a message first.")
        return

    response = requests.post(
        f"http://127.0.0.1:8000/tickets/{ticket_id}/reply",
        json={
            "message": message
        }
    )

    if response.status_code == 200:

        agent_input.value = ""
        load_tickets()
        load_conversation()

    else:

        print(response.text)


def resolve_ticket(button):

    ticket_id = ticket_dropdown.value

    if not ticket_id:
        print("Select a ticket first.")
        return

    response = requests.post(
        f"http://127.0.0.1:8000/tickets/{ticket_id}/resolve"
    )

    if response.status_code == 200:

        print("Ticket resolved successfully.")

        load_tickets()
        load_conversation()

    else:

        print(response.text)


ticket_dropdown.observe(
    load_conversation,
    names="value"
)

send_button.on_click(send_message)

resolve_button.on_click(resolve_ticket)

load_tickets()


display(
    widgets.HTML("<h2>Human Agent Dashboard</h2>"),
    ticket_dropdown,
    conversation_box,
    agent_input,
    widgets.HBox([
        send_button,
        resolve_button
    ])
)
```

```python
import requests
import ipywidgets as widgets
from IPython.display import display

session_id = "customer_test_001"

chat_box = widgets.Output(
    layout=widgets.Layout(
        border="1px solid #ccc",
        padding="10px",
        height="350px",
        overflow="auto"
    )
)

message_input = widgets.Text(
    placeholder="Type your message...",
    layout=widgets.Layout(width="70%")
)

send_button = widgets.Button(
    description="Send",
    button_style="primary"
)

refresh_button = widgets.Button(
    description="Refresh",
    button_style="info"
)

ticket_id = None
```

```python
def show_messages():

    chat_box.clear_output(wait=True)

    if not ticket_id:
        return

    response = requests.get(
        f"http://127.0.0.1:8000/tickets/{ticket_id}/messages"
    )

    if response.status_code != 200:
        with chat_box:
            print("Unable to load conversation.")
        return

    data = response.json()

    with chat_box:

        for msg in data["messages"]:

            if msg["sender"] == "customer":
                print(f"You: {msg['message']}")

            elif msg["sender"] == "agent":
                print(f"Human Agent: {msg['message']}")

            print("-" * 60)
```

```python
def send_message(button):

    global ticket_id

    message = message_input.value.strip()

    if not message:
        return

    if ticket_id:

        response = requests.post(
            f"http://127.0.0.1:8000/tickets/{ticket_id}/customer-reply",
            json={
                "message": message
            }
        )

    else:

        response = requests.post(
            "http://127.0.0.1:8000/chat",
            json={
                "session_id": session_id,
                "message": message
            }
        )

    if response.status_code != 200:

        with chat_box:
            print("Error:", response.text)

        return

    data = response.json()

    if data.get("ticket_id"):
        ticket_id = data["ticket_id"]

    message_input.value = ""

    if ticket_id:
        show_messages()

    else:
        with chat_box:
            print(f"You: {message}")
            print(f"ACE Helper: {data['response']}")
            print("-" * 60)
```

```python
def refresh_messages(button):
    show_messages()


send_button.on_click(send_message)
refresh_button.on_click(refresh_messages)


display(
    widgets.HTML(
        "<h2>ACE Helper - Customer Support</h2>"
    ),
    chat_box,
    widgets.HBox([
        message_input,
        send_button,
        refresh_button
    ])
)
```

```python
import json

with open("Ace_Helper.ipynb", "r", encoding="utf-8") as f:
    notebook = json.load(f)

keep_cells = []

keep_cells.append({
    "cell_type": "markdown",
    "metadata": {},
    "source": [
        "# ACE Helper\n",
        "\n",
        "AI-powered Customer Support & Ticketing System\n",
        "\n",
        "ACE Helper automatically handles customer queries, retrieves order information, and escalates unresolved issues to human support agents."
    ]
})

keep_cells.append({
    "cell_type": "markdown",
    "metadata": {},
    "source": [
        "## System Flow\n",
        "\n",
        "Customer → AI Assistant → Answer / Order Lookup / Human Escalation → Ticket → Human Agent → Resolution"
    ]
})

for index in range(242, 249):
    if index < len(notebook["cells"]):
        keep_cells.append(notebook["cells"][index])

new_notebook = {
    "cells": keep_cells,
    "metadata": notebook.get("metadata", {}),
    "nbformat": notebook.get("nbformat", 4),
    "nbformat_minor": notebook.get("nbformat_minor", 5)
}

with open("ACE_Helper_Final.ipynb", "w", encoding="utf-8") as f:
    json.dump(new_notebook, f, indent=1)

print("ACE_Helper_Final.ipynb created successfully.")
print("Cells:", len(keep_cells))
```
